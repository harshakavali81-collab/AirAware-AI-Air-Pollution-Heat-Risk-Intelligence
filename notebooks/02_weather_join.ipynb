# 02 · Weather join (source-separation example)

**SYNTHETIC DEMONSTRATION DATA — NOT REAL CITY MEASUREMENTS.**

```python
from src.airaware.ingestion.demo import demo_readings
from src.airaware.processing.features import daily_city_average,join_weather
import pandas as pd
```

```python
pm=daily_city_average(demo_readings('Hyderabad',30))
```

```python
weather=pd.DataFrame({'date':pd.to_datetime(pm.date).dt.date.astype(str),'temperature_c':[29+i%8 for i in range(len(pm))]})
# Generated *example weather*, not fetched from a weather service.
joined=join_weather(pm,weather)
joined.tail()
```

```python
print('Missing values:',joined.isna().sum().to_dict())
```
