# 01 · PM2.5 exploratory analysis

**SYNTHETIC DEMONSTRATION DATA — NOT REAL CITY MEASUREMENTS.**

```python
from src.airaware.ingestion.demo import demo_readings
from src.airaware.processing.validation import validate_measurements, coverage
import matplotlib.pyplot as plt
```

```python
raw=demo_readings('Hyderabad',120)
df=validate_measurements(raw)
print(coverage(df,120))
df.head()
```

```python
daily=df.groupby('date',as_index=False).pm25.mean()
ax=daily.plot(x='date',y='pm25',title='SYNTHETIC PM2.5 daily city mean',figsize=(11,4))
ax.set_ylabel('µg/m³'); plt.show()
```
